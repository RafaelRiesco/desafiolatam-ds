3. Revisión de outliers: Ahora que tienes variables numéricas revisa la distribución y
utiliza algún método para encontrar outliers, por ejemplo IQR o Z-score.

```python
# Detección de outliers con IQR en variables numéricas clave
def detectar_outliers_iqr(df, columna, factor=1.5):
    Q1 = df[columna].quantile(0.25)
    Q3 = df[columna].quantile(0.75)
    IQR = Q3 - Q1
    limite_inf = Q1 - factor * IQR
    limite_sup = Q3 + factor * IQR
    outliers = df[(df[columna] < limite_inf) | (df[columna] > limite_sup)]
    return outliers, limite_inf, limite_sup

print('=== Resumen de Outliers (IQR, factor=1.5) ===\n')
for col in ['User Rating', 'Number of Votes', 'Runtime_num', 'Year_num']:
    outliers, lim_inf, lim_sup = detectar_outliers_iqr(df_base.dropna(subset=[col]), col)
    print(f'{col}:')
    print(f'  Rango válido: [{lim_inf:.1f}, {lim_sup:.1f}]')
    print(f'  Outliers: {len(outliers)} ({len(outliers)/len(df_base)*100:.1f}%)')
    print()
```

4. Transformación de variables finales: Realiza un pequeño análisis de distribuciones y
transforma las variables aplicando transformaciones como logaritmo o
get_dummies para extraer las diferentes categorías.
Genera una estrategia para lidiar con los valores nulos y crea las variables que te
parezcan necesarias.

```python
# 1. Number of Votes — transformación logarítmica
df_base['votes_log'] = np.log1p(df_base['Number of Votes'])

# 2. Runtime — agrupar en rangos
def clasificar_runtime(mins):
    if pd.isna(mins): return np.nan
    if mins <= 5: return 'Short'        # cortometrajes
    if mins <= 30: return 'Episode'     # episodios TV típicos
    if mins <= 60: return 'Medium'      # episodios largos
    if mins <= 120: return 'Movie'      # películas
    return 'Long'                        # contenido muy largo

df_base['Runtime_cat'] = df_base['Runtime_num'].apply(clasificar_runtime)
print('Distribución Runtime_cat:')
print(df_base['Runtime_cat'].value_counts())
```

```python
# 3. Year — agrupar en décadas
def clasificar_decada(year):
    if pd.isna(year): return np.nan
    if year < 1980: return 'Pre-1980'
    if year < 1990: return '1980s'
    if year < 2000: return '1990s'
    if year < 2010: return '2000s'
    if year < 2020: return '2010s'
    return '2020s'

df_base['Decade'] = df_base['Year_num'].apply(clasificar_decada)
print('\nDistribución por Década:')
print(df_base['Decade'].value_counts().sort_index())
```

```python
# 4. Estrategia para nulos restantes
print('\n=== Nulos restantes en variables clave ===')
vars_clave = ['User Rating', 'votes_log', 'Runtime_cat', 'Year_num', 'Certificate_group', 'Decade']
print(df_base[vars_clave].isnull().sum())
```

```python
# ESTRATEGIA PARA NULOS

# Runtime_cat: imputar con 'Unknown' — no podemos inferir duración
# Certificate_group: imputar con 'Unrated' — sin clasificación conocida

df_base['Runtime_cat'] = df_base['Runtime_cat'].fillna('Unknown')
df_base['Certificate_group'] = df_base['Certificate_group'].fillna('Unrated')

print('Nulos después de imputar:')
print(df_base[['Runtime_cat', 'Certificate_group']].isnull().sum())
print()
print('Distribución Runtime_cat final:')
print(df_base['Runtime_cat'].value_counts())
print()
print('Distribución Certificate_group final:')
print(df_base['Certificate_group'].value_counts())
```

```python
# Get dummies para Certificate_group, Runtime_cat y Decade
cert_dummies = pd.get_dummies(df_base['Certificate_group'], prefix='cert')
runtime_dummies = pd.get_dummies(df_base['Runtime_cat'], prefix='runtime')
decade_dummies = pd.get_dummies(df_base['Decade'], prefix='decade')

# Agregar al dataset
df_base = pd.concat([df_base, cert_dummies, runtime_dummies, decade_dummies], axis=1)

print(f'Dimensiones finales: {df_base.shape}')
print(f'\nNuevas columnas agregadas:')
print('Certificate:', cert_dummies.columns.tolist())
print('Runtime:', runtime_dummies.columns.tolist())
print('Decade:', decade_dummies.columns.tolist())
```

```python
# ANOVA: User Rating vs variables categóricas
from scipy.stats import f_oneway

print('=== Test ANOVA: User Rating vs variables categóricas ===\n')

for var_cat in ['Certificate_group', 'Runtime_cat', 'Decade']:
    grupos = [
        df_base[df_base[var_cat] == cat]['User Rating'].dropna()
        for cat in df_base[var_cat].unique()
    ]
    stat, p_value = f_oneway(*grupos)
    significativo = '✓ Significativo' if p_value < 0.05 else '✗ No significativo'
    print(f'{var_cat}: F={stat:.2f}, p-value={p_value:.4f} → {significativo}')
```

```python
def preprocesar_anime(filepath):
    """
    Lee el dataset original de anime IMDB y retorna un dataset
    completamente procesado con todas las transformaciones aplicadas.
    """
    # Carga
    df = pd.read_excel(filepath)

    # Eliminar columnas no útiles
    df = df.drop(columns=['Metascore', 'Summary', 'Episode Title', 'Stars'])

    # Eliminar fila con header duplicado y duplicados exactos
    df = df[df['Episode'] != 'Episode']
    df = df.drop_duplicates()

    # Convertir Episode y filtrar con User Rating y Number of Votes
    df['Episode'] = pd.to_numeric(df['Episode'], errors='coerce')
    df = df.dropna(subset=['Episode'])
    df = df[df['User Rating'].notna() & df['Number of Votes'].notna()].copy()

    # Convertir numéricas
    df['User Rating'] = pd.to_numeric(df['User Rating'], errors='coerce')
    df['Number of Votes'] = pd.to_numeric(df['Number of Votes'], errors='coerce')

    # Eliminar Gross (idéntico a Number of Votes)
    df = df.drop(columns=['Gross'])

    # Transformar Year
    df['Year_num'] = df['Year'].str.extract(r'\((\d{4})')[0].astype(float)
    mask_nulos_year = df['Year_num'].isna()
    df.loc[mask_nulos_year, 'Year_num'] = pd.to_numeric(
        df.loc[mask_nulos_year, 'Year'], errors='coerce').abs()

    # Transformar Runtime
    df['Runtime_num'] = df['Runtime'].str.extract(r'(\d+)').astype(float)

    # Agrupar Runtime en categorías
    def clasificar_runtime(mins):
        if pd.isna(mins): return 'Unknown'
        if mins <= 5: return 'Short'
        if mins <= 30: return 'Episode'
        if mins <= 60: return 'Medium'
        if mins <= 120: return 'Movie'
        return 'Long'
    df['Runtime_cat'] = df['Runtime_num'].apply(clasificar_runtime)

    # Agrupar Year en décadas
    def clasificar_decada(year):
        if pd.isna(year): return 'Unknown'
        if year < 1980: return 'Pre-1980'
        if year < 1990: return '1980s'
        if year < 2000: return '1990s'
        if year < 2010: return '2000s'
        if year < 2020: return '2010s'
        return '2020s'
    df['Decade'] = df['Year_num'].apply(clasificar_decada)

    # Agrupar Certificate
    cert_map = {
        'TV-G': 'All', 'G': 'All', 'TV-Y': 'All', 'U': 'All',
        'Uc': 'All', 'E': 'All', 'Open': 'All',
        'TV-Y7': 'Kids', 'TV-Y7-FV': 'Kids', 'E10+': 'Kids',
        'TV-PG': 'PG', 'PG': 'PG', 'TV-13': 'PG', '12': 'PG',
        '12A': 'PG', '13+': 'PG',
        'TV-14': 'Teen', 'PG-13': 'Teen', 'T': 'Teen', '15': 'Teen',
        '16+': 'Teen', 'M': 'Teen',
        'TV-MA': 'Adult', 'R': 'Adult', '18': 'Adult', '18+': 'Adult',
        'R18': 'Adult', 'X': 'Adult',
        'Not Rated': 'Unrated', 'Unrated': 'Unrated',
        'Approved': 'Unrated', 'Banned': 'Unrated'
    }
    df['Certificate_group'] = df['Certificate'].map(cert_map).fillna('Unrated')

    # Transformación logarítmica de votos
    df['votes_log'] = np.log1p(df['Number of Votes'])

    # Get dummies
    genre_dummies = df['Genre'].str.split(', ').str.join('|').str.get_dummies()
    cert_dummies = pd.get_dummies(df['Certificate_group'], prefix='cert')
    runtime_dummies = pd.get_dummies(df['Runtime_cat'], prefix='runtime')
    decade_dummies = pd.get_dummies(df['Decade'], prefix='decade')

    df = pd.concat([df, genre_dummies, cert_dummies, runtime_dummies, decade_dummies], axis=1)

    print(f'Dataset procesado: {df.shape}')
    print(f'Nulos restantes en variables clave:')
    vars_clave = ['User Rating', 'votes_log', 'Year_num', 'Runtime_cat',
                  'Certificate_group', 'Decade']
    print(df[vars_clave].isnull().sum())

    return df

# Probar la función
df_procesado = preprocesar_anime('imdb_anime.xlsx')
df_procesado.head()
```

7. A partir de las columnas que obtuviste realiza una selección de variables según los
siguientes métodos:
a. Filtros basados en correlaciones
b. Forward Selection.

```python
# Variables features disponibles (excluir texto original y variable objetivo)
cols_excluir = ['Title', 'Genre', 'User Rating', 'Number of Votes',
                'Runtime', 'Year', 'Certificate', 'Runtime_num',
                'Episode', 'Episode Title', 'Summary', 'Stars']

X = df_procesado.drop(columns=[c for c in cols_excluir if c in df_procesado.columns])
y = df_procesado['User Rating']

print(f'Features disponibles: {X.shape[1]}')
print(X.columns.tolist())
```

```python
# Mantener solo columnas numéricas para los modelos
X_num = X.select_dtypes(include=[np.number])
print(f'Features numéricas: {X_num.shape[1]}')
print(X_num.columns.tolist())
```

```python
# Seleccionar features numéricas + dummies de categorías
cols_dummies = [c for c in df_procesado.columns if c.startswith(('cert_', 'runtime_', 'decade_'))]
X_final = pd.concat([X_num, df_procesado[cols_dummies]], axis=1)

print(f'Features totales para modelado: {X_final.shape[1]}')
print(X_final.columns.tolist())
```

```python
# ── Método 1: Filtro por correlaciones ──────────────────────
# Eliminamos features con correlación > 0.7 entre sí (multicolinealidad)
# y mantenemos las más correlacionadas con User Rating

corr_matrix = X_final.corr().abs()

# Identificar pares con alta correlación entre features
upper = corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))
to_drop_corr = [col for col in upper.columns if any(upper[col] > 0.7)]

print(f'Features eliminadas por alta correlación entre sí: {len(to_drop_corr)}')
print(to_drop_corr)

X_filtrado = X_final.drop(columns=to_drop_corr)
print(f'\nFeatures restantes: {X_filtrado.shape[1]}')
print(X_filtrado.columns.tolist())
```

```python
# ── Método 2: Forward Selection ─────────────────────────────
from sklearn.feature_selection import SequentialFeatureSelector
from sklearn.linear_model import LinearRegression

# Usamos regresión lineal como modelo base
# n_features_to_select=10 para comparar top features con correlación
modelo_base = LinearRegression()

sfs = SequentialFeatureSelector(
    modelo_base,
    n_features_to_select=10,
    direction='forward',
    scoring='r2',
    cv=5
)

sfs.fit(X_filtrado, y)

features_forward = X_filtrado.columns[sfs.support_].tolist()
print('Features seleccionadas por Forward Selection (top 10):')
for i, f in enumerate(features_forward, 1):
    print(f'  {i}. {f}')
```

```python
# ── Comparación de métodos ───────────────────────────────────

# Top 10 por correlación con User Rating
top_corr = X_final.corrwith(y).abs().sort_values(ascending=False).head(10)

print('=== Comparación de métodos de selección ===\n')
print('Top 10 por CORRELACIÓN con User Rating:')
for i, (feat, val) in enumerate(top_corr.items(), 1):
    en_forward = '✓' if feat in features_forward else '✗'
    print(f'  {i}. {feat:<25} r={val:.3f}  Forward: {en_forward}')

print()
print('Seleccionadas por FORWARD SELECTION:')
for i, feat in enumerate(features_forward, 1):
    corr_val = abs(X_final.corrwith(y)[feat])
    en_top_corr = '✓' if feat in top_corr.index else '✗'
    print(f'  {i}. {feat:<25} r={corr_val:.3f}  Top Corr: {en_top_corr}')
```

Comparación de métodos
Coincidencias (6 de 10): cert_Unrated, Short, runtime_Short, decade_Pre-1980, runtime_Episode, runtime_Movie — ambos métodos las seleccionaron.

Solo en Correlación: votes_log, cert_Teen, Action, Adventure — variables con correlación individual alta pero que Forward Selection no eligió, posiblemente porque su información ya está capturada por otras variables del conjunto.

Solo en Forward Selection: Sci-Fi, cert_All, runtime_Medium, decade_1990s — variables con correlación individual baja (0.08-0.12) pero que en combinación con las otras mejoran el R² del modelo.

Ambos métodos coinciden en 6 de 10 variables, lo que refuerza
su relevancia como predictores de User Rating.

Filtro por correlación es más simple e interpretable — prioriza
variables con relación directa con el objetivo. Sus top variables
(votes_log, cert_Unrated, Short) coinciden con los hallazgos
del análisis exploratorio: más votos = mejor rating, contenido
sin clasificar = peor rating, cortometrajes = peor rating.

Forward Selection captura interacciones entre variables — elige
Sci-Fi y cert_All que individualmente tienen baja correlación,
pero en combinación con runtime y decade mejoran el modelo.

Diferencia clave: votes_log (r=0.290, la más correlacionada)
NO fue seleccionada por Forward Selection, probablemente porque
su efecto ya está capturado por runtime_Episode y las variables
de certificación que juntas explican mejor la varianza.

Recomendación: usar el conjunto de Forward Selection para modelado
(considera interacciones), y el de correlación para interpretación
de negocio (más intuitivo).
